## 3. (i)

```python
# Building the KNN Prediction Model

import numpy as np

def knn_predict(X_train, y_train, X_test, k):
    distances = []

    # Calculate Euclidean distances
    for i in range(len(X_train)):
        distance = np.sqrt(np.sum((X_train[i] - X_test)**2))
        distances.append(distance)

    # Sort distances and get indices
    sorted_index = np.argsort(distances)

    # Select K nearest indices
    nearest_index = sorted_index[:k]

    # Get classes of nearest neighbors
    nearest_classes = y_train[nearest_index]
    classes, counts = np.unique(nearest_classes, return_counts = True)
    pred = classes[np.argmax(counts)]

    return pred
```

```python
X_train = np.array([[2.0, 60.0],
                    [3.0, 65.0],
                    [4.0, 70.0],
                    [5.0, 75.0],
                    [6.0, 80.0]])
y_train = np.array(['Fail', 'Fail', 'Pass', 'Pass', 'Pass'])

X_test = np.array([3.5, 68.0])
k = 3

final_prediction = knn_predict(X_train, y_train, X_test, k)
print(f"The model predicts Student X will: {final_prediction}")
```

## 3.(iii)

```python
import numpy as np

def knn_regression(X_train, y_train, X_test, k):

    distances = []

    # Calculate Euclidean distances
    for i in range(len(X_train)):
        distance = np.sqrt(np.sum((X_train[i] - X_test)**2))
        distances.append(distance)

    # Sort distances and get indices
    sorted_index = np.argsort(distances)

    # Get indices of K nearest neighbors
    nearest_index = sorted_index[:k]

    # Get consumption values of K nearest neighbors
    nearest_values = y_train[nearest_index]

    # Calculate average consumption
    pred = np.mean(nearest_values)

    return pred
```

```python
X_train = np.array([
    [2, 8],
    [3, 10],
    [3, 12],
    [4, 14],
    [5, 16]
])
y_train = np.array([
    180,
    220,
    250,
    280,
    320
])

X_test = np.array([3, 11])
k = 3

pred = knn_regression(X_train, y_train, X_test, k)
print(f"Predicted Electricity Consumption: {pred:.2f} kWh")
```

```python
# Creating the dataframe and saving as CSV file
import pandas as pd

data = {
    "Age": [25, 40, 35, 28, 50, 23, 45, 33, 29, 38],
    "Income": [30000, 60000, 45000, 32000, 80000,25000, 70000, 48000, 35000, 52000],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female"],
    "Education": ["Graduate", "Postgrad", "Graduate", "School", "Postgrad", "Graduate", "Graduate", "Postgrad", "School", "Graduate"],
    "Credit_Score": [650, 720, 680, 600, 750, 620, 710, 690, 640, 705],
    "Loan_Approved": ["No", "Yes", "Yes", "No", "Yes", "No", "Yes", "Yes", "No", "Yes"]
}
df = pd.DataFrame(data)
df.to_csv("loan_data.csv", index=False)

print("CSV file created successfully!")
```

```python
# Reading the CSV file
df = pd.read_csv("loan_data.csv")
display(df)
```

```python
# Importing required libraries
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Encoding the categorical data
le_gender, le_education, le_loan = LabelEncoder(), LabelEncoder(), LabelEncoder()

df["Gender"] = le_gender.fit_transform(df["Gender"])
df["Education"] = le_education.fit_transform(df["Education"])
df["Loan_Approved"] = le_loan.fit_transform(df["Loan_Approved"])

display(df)
```

```python
# Separating features and target
X = df.drop("Loan_Approved", axis=1)
y = df["Loan_Approved"]
print("Features:")
display(X)
print("\nTarget:")
print(y)
```

```python
# Scaling the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Training KNNN

knn = KNeighborsClassifier(n_neighbors=3, metric="euclidean")
knn.fit(X_scaled, y)
print("KNN model trained successfully!")
```

```python
new_data = pd.DataFrame({
    "Age": [30, 42, 26, 48, 36],
    "Income": [40000, 65000, 28000, 75000, 50000],
    "Gender": ["Male", "Female", "Female", "Male", "Female"],
    "Education": ["Graduate", "Postgrad", "School", "Postgrad", "Graduate"],
    "Credit_Score": [670, 730, 610, 740, 695]
})
display(new_data)
```

```python
new_data_original = new_data.copy()
```

```python
# encoding the new instances
new_data["Gender"] = le_gender.transform(new_data["Gender"])
new_data["Education"] = le_education.transform(new_data["Education"])
```

```python
# scaling
new_data_scaled = scaler.transform(new_data)
```

```python
predictions = knn.predict(new_data_scaled)
predicted_labels = le_loan.inverse_transform(predictions)
```

```python
new_data_original["Predicted_Loan_Approved"] = predicted_labels
display(new_data_original)
```

```python
import pandas as pd
data = {
    "Area": [1200, 2000, 1500, 800, 2200, 1700, 900, 1400, 2500, 1600],
    "Bedrooms": [2, 3, 3, 1, 4,3, 2, 2, 4, 3],
    "Age": [10, 5, 8, 15, 3, 6, 12, 7, 2, 9],
    "Location": ["Urban", "Urban", "Suburban", "Rural", "Urban", "Suburban", "Rural", "Urban", "Urban", "Suburban"],
    "Condition": ["Good", "Excellent", "Good", "Average", "Excellent", "Good", "Average", "Good", "Excellent", "Good"],
    "Price": [30.5, 55.0, 42.0, 18.0, 65.5, 48.0, 20.0, 38.5, 75.0, 45.0]
}

df = pd.DataFrame(data)
df.to_csv("house_data.csv", index=False)
print("CSV file created successfully!")
```

```python
df = pd.read_csv("house_data.csv")
print("\nOriginal Dataset:")
display(df)
```

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.neighbors import KNeighborsRegressor

le_location, le_condition = LabelEncoder(), LabelEncoder()

df["Location"] = le_location.fit_transform(df["Location"])
df["Condition"] = le_condition.fit_transform(df["Condition"])

print("\nEncoded Dataset:")
display(df)
```

```python
X = df.drop("Price", axis=1)
y = df["Price"]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
knn = KNeighborsRegressor(
    n_neighbors=3,
    metric="euclidean"
)
knn.fit(X_scaled, y)
print("KNN Regression Model Trained Successfully!")
```

```python
new_data = pd.DataFrame({
    "Area": [1300, 2100, 1000, 1800, 2400],
    "Bedrooms": [2, 3, 2, 3, 4],
    "Age": [8, 4, 10, 6, 3],
    "Location": ["Urban", "Urban", "Rural", "Suburban", "Urban"],
    "Condition": ["Good", "Excellent", "Average", "Good", "Excellent"]
})

new_data_original = new_data.copy()
```

```python
new_data["Location"] = le_location.transform(new_data["Location"])
new_data["Condition"] = le_condition.transform(new_data["Condition"])

new_data_scaled = scaler.transform(new_data)
```

```python
predictions = knn.predict(new_data_scaled)
new_data_original["Predicted Price (Lakhs)"] = predictions

print("\nPrediction Results:")
display(new_data_original)
```
